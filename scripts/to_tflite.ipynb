```python
import tensorflow as tf

model = tf.keras.models.load_model('../models/eye_model_finetuned.h5')  # <--- .h5 model file path

converter = tf.lite.TFLiteConverter.from_keras_model(model)
tflite_model = converter.convert()

with open('../models/eye_model.tflite', 'wb') as f:   # <--- .tflite model file path
    f.write(tflite_model)
```
